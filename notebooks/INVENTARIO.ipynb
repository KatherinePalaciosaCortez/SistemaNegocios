```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/KatherinePalaciosaCortez/SistemaNegocios/refs/heads/main/data/raw/global_ecommerce_sales.csv"
df = pd.read_csv(url, sep=';')

df.head()
```

```python
#Valores nulos
print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
#valores duplicados
print("\nCantidad de registros duplicados:")
print(df.duplicated().sum())
```

```python
# tipos de datos
print("\nInformación general del dataset:")
print(df.info())
```

```python
#  Validamos las dimensiones del dataset
print("Dimensiones del dataset:")
print(df.shape)

```

```python
#  Revisamos las columnas disponibles
print("\nColumnas del dataset:")
print(df.columns)

```

```python
# Mostrar las primeras filas
print("\nPrimeras filas del dataset:")
print(df.head())
```

```python

#convertir las fechas
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    dayfirst=True
)
```

```python
print(df["Order_Date"].dtype)
```

```python
#crear calendariodf
df["Año"] = df["Order_Date"].dt.year

df["Mes"] = df["Order_Date"].dt.month

df["Nombre_Mes"] = df["Order_Date"].dt.month_name()

df["Trimestre"] = df["Order_Date"].dt.quarter

df["Dia"] = df["Order_Date"].dt.day
```

```python
#limpieza de texto
columnas_texto = [
    "Customer_Name",
    "Customer_Segment",
    "Country",
    "Region",
    "Product_Category",
    "Product_Name",
    "Payment_Method"
]

for col in columnas_texto:
    df[col] = df[col].str.strip()
    df
```

```python
#metricas de negocio
#margen%
df["Margen_Porcentaje"] = (
    df["Profit"] /
    df["Total_Sales"]
) * 100
```

```python
#venta neta
df["Venta_Neta"] = (
    df["Total_Sales"] -
    df["Shipping_Cost"]
)
```

```python
#descuento aplicado
df["Monto_Descuento"] = (
    df["Quantity"] *
    df["Unit_Price"]
) * (
    df["Discount_Percent"] / 100
)
```

```python
#kpi ventas
ventas_totales = df["Total_Sales"].sum()
```

```python
#clientes unicos
clientes = df["Customer_Name"].nunique()
```

```python
utilidad_total = df["Profit"].sum()
#margen promedio
margen = (
    utilidad_total /
    ventas_totales
) * 100
```

```python

#ventas por mes
ventas_mes = (
    df.groupby(["Año","Mes"])
    ["Total_Sales"]
    .sum()
    .reset_index()
)
```

```python
#ventas por region

ventas_region = (
    df.groupby("Region")
    ["Total_Sales"]
    .sum()
    .sort_values(
        ascending=False
    )
)

```

```python
#ventas por pais
ventas_pais = (
    df.groupby("Country")
    ["Total_Sales"]
    .sum()
    .sort_values(
        ascending=False
    )
)
```
